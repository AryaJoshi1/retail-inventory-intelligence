```python
# Inspect the Sales Dataset
sales_df = spark.table("workspace.default.retail_sales_ml_apl")

print("Sales row count:", sales_df.count())

sales_df.printSchema()

display(sales_df.limit(10))
```

```python
# Inspect the Inventory Dataset
inventory_df = spark.table("workspace.default.retail_inventory_ml_apl")

print("Inventory row count:", inventory_df.count())

inventory_df.printSchema()

display(inventory_df.limit(10))
```

```python
# Basic data profiling

print("========== SALES ==========")

print("Rows:", sales_df.count())
print("Columns:", len(sales_df.columns))

print("\nDate range:")
sales_df.selectExpr(
    "min(`Transaction Date`) as min_date",
    "max(`Transaction Date`) as max_date"
).show()

print("\nUnique products:", sales_df.select("`Product No`").distinct().count())
print("Unique stores:", sales_df.select("Store").distinct().count())
print("Unique suppliers:", sales_df.select("Supplier").distinct().count())
print("Unique sales channels:", sales_df.select("`Sales Channel`").distinct().count())


print("\n========== INVENTORY ==========")

print("Rows:", inventory_df.count())
print("Columns:", len(inventory_df.columns))

print("\nDate range:")
inventory_df.selectExpr(
    "min(`Start Date`) as min_date",
    "max(`End Date`) as max_date"
).show()

print("\nUnique products:", inventory_df.select("`Product No`").distinct().count())
print("Unique stores:", inventory_df.select("Store").distinct().count())
print("Unique suppliers:", inventory_df.select("Supplier").distinct().count())
print("Unique sales channels:", inventory_df.select("`Sales Channel`").distinct().count())
```

```python
# Examine important categorical fields

print("SALES TYPES")
display(
    sales_df.groupBy("`Sales Type`")
    .count()
    .orderBy("count", ascending=False)
)

print("RETURNS")
display(
    sales_df.groupBy("`Is Return`")
    .count()
)

print("INVENTORY STOCK STATUS")
display(
    inventory_df.groupBy("`Stock Status`")
    .count()
    .orderBy("count", ascending=False)
)
```

```python
# Check missing values

from pyspark.sql.functions import col, sum as spark_sum

print("SALES NULLS")

sales_nulls = sales_df.select([
    spark_sum(col(c).isNull().cast("int")).alias(c)
    for c in sales_df.columns
])

display(sales_nulls)

print("INVENTORY NULLS")

inventory_nulls = inventory_df.select([
    spark_sum(col(c).isNull().cast("int")).alias(c)
    for c in inventory_df.columns
])

display(inventory_nulls)
```

```python
# Check duplicate records

print("Sales duplicates:")
print(
    sales_df.count() -
    sales_df.dropDuplicates().count()
)

print("\nInventory duplicates:")
print(
    inventory_df.count() -
    inventory_df.dropDuplicates().count()
)
```

```python
# Check Product + Store overlap

sales_keys = sales_df.select(
    "`Product No`", "Store"
).distinct()

inventory_keys = inventory_df.select(
    "`Product No`", "Store"
).distinct()

print("Sales Product-Store combinations:",
      sales_keys.count())

print("Inventory Product-Store combinations:",
      inventory_keys.count())

print("Sales combinations missing from Inventory:",
      sales_keys.join(
          inventory_keys,
          ["Product No", "Store"],
          "left_anti"
      ).count())
```
